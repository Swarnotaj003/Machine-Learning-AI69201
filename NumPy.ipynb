# NumPy

NumPy is a Python library for numerical computing that provides high-performance multidimensional arrays and is implemented primarily in C for efficient execution.

```python
import numpy as np
```

## Array

Linear data structure

```python
lst = [i for i in range(1, 6)]
arr = np.array(lst)
type(arr)
```

```python
arr
```

```python
print(arr.shape)
print(arr)
```

## Multi-nested Array

Array of arrays

```python
lst1 = [i for i in range(6, 11)]
lst2 = [i**2 for i in range(6, 11)]
lst3 = [i**3 for i in range(6, 11)]

mat = np.array([lst1, lst2, lst3])
type(mat)
```

```python
mat
```

```python
print(mat.shape)
print(mat)
```

## Reshape

The product of input rows and columns must be equal to the number of elements in the original array

```python
mat1 = mat.reshape(5, 3)
mat1
```

```python
mat1.shape
```

## Copying Array

Broadcasting

```python
print(arr)
arr[2:] = 100
print(arr)
```

Deep copy (Array is a reference type)

```python
arr1 = arr
arr1[3:] = 500
print(arr1)
print(arr)
```

Shallow copy

```python
arr2 = arr.copy()
arr2[3:] = 999
print(arr2)
print(arr)
```

## In-built Functions

### arange()

```python
lst1to10 = np.arange(1, 10)
lst1to10
```

```python
lst1to10step2 = np.arange(1, 10, 3)
lst1to10step2
```

### linspace()

```python
linSpace = np.linspace(1, 10, 25)
linSpace
```

### ones()

```python
ones1 = np.ones(4)
ones1
```

```python
ones2 = np.ones((2, 5), dtype=int)
ones2
```

### rand()
Random distribution over [0, 1)

```python
rd = np.random.rand(3, 3)
rd
```

### randn()
Standard Normal distribution

```python
snd = np.random.randn(2, 4)
snd
```

### randint()

```python
rint = np.random.randint(1, 100, 15).reshape(3, 5)
rint
```

### random_sample()

Same as rand(), but takes array shape as a tuple input

```python
rsamp = np.random.random_sample((2, 3))
rsamp
```
