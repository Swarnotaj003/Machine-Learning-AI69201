# MatplotLib

Matplotlib is a plotting library for Python and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits.

```python
import numpy as np
import matplotlib.pyplot as plt
# to avoid using plot.show() in every cell of Jupyter notebook, use
# %matplotlib inline 
```

```python
x = np.arange(0, 10)
y = np.arange(11, 21)
```

```python
a = np.arange(40, 50)
b = np.arange(50, 60)
```

## Scatter Plot

```python
plt.scatter(x, y, c = 'r')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('My First Graph!!')
plt.savefig('images/my_first_graph.png')
```

## Plot

```python
y = x**2
# solid line
plt.plot(x, y, 'g')     
```

```python
# round points
plt.plot(x, y, 'go')
```

```python
# round points with dashed line
plt.plot(x, y, 'go--')
```

```python
# star points with solid line
plt.plot(x, y, 'g*-')
```

## Subplots

```python
# create a canvas of size 2x3 and place the plot on 2nd position
plt.subplot(2, 3, 2)
plt.plot(x, x, 'r--')
plt.title('Identity')

plt.subplot(2, 3, 4)    # 4th position
plt.plot(x, x**2, 'g*')
plt.title('Quadratic')

plt.subplot(2, 3, 5)    # 5th position
plt.plot(x, 3*x + 5, 'y*-')
plt.title('Linear scale & shift')

plt.subplot(2, 3, 6)    # 6th position
plt.plot(x, x**3, 'bo--')
plt.title('Cubic')
```

```python
x = np.arange(0, 4 * np.pi, 0.1)    # use smaller step value for smoother curve
y_sin = np.sin(x)                   # sine wave
y_cos = np.cos(x)                   # cosine wave

plt.subplot(2, 1, 1)
plt.plot(x, y_sin, c = 'purple')
plt.title('Sine')

plt.subplot(2, 1, 2)
plt.plot(x, y_cos, c = 'orange')
plt.title('Cosine')
```

## Bar Plot

```python
x1 = [2, 8, 10]
y1 = [11, 16, 9]

x2 = [3, 9, 11]
y2 = [6, 15, 7]

plt.bar(x1, y1)
plt.bar(x2, y2)
```

## Histogram

```python
arr = np.array([22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27])
plt.hist(arr)
```

## Box Plot

```python
data = [np.random.normal(0, std, 100) for std in range(1,4)]
plt.boxplot(data, vert = True, patch_artist = True)
```

## Pie Chart

```python
labels = ['Python', 'C++', 'Ruby', 'Java']
sizes = [240, 160, 80, 320]
colors = ['gold', 'lime', 'coral', 'skyblue']
explode = [0.2, 0, 0, 0]    # Explode 1st slice

plt.pie(sizes, explode, labels, colors, shadow = True, autopct = '%.2f%%')
plt.axis('equal')   # ensures equal scaling of x and y axes
```
