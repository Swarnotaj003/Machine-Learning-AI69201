# Seaborn

Seaborn is a high-level data visualization library for Python built on top of Matplotlib. It provides a simple interface for creating attractive and informative statistical graphics and integrates seamlessly with pandas DataFrames.

```python
import seaborn as sns
```

```python
# Seaborn has its own sample datasets
df = sns.load_dataset('tips')
df.head()
```

```python
df.info()
```

## Correlation with Heatmap

A correlation heatmap uses colored cells, typically in a monochromatic scale, to show a 2D correlation matrix between two discrete dimensions.

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

```python
sns.heatmap(corr_matrix)
```

## Join Plot

A join plot allows to study the relationship between two numeric variables. The central chart display their correlation. It is usually a scatterplot, a hexbin plot, a 2D histogram or a 2D density plot.


```python
# with hex bins
sns.jointplot(data=df, x='total_bill', y='tip', kind='hex', color='green')
```

```python
# with regression line and PDF
sns.jointplot(data=df, x='total_bill', y='tip', kind='reg', color='magenta')
```

## Pair Plot

A pair plot is used to visualize pairwise relationships among multiple numerical variables in a dataset. It displays scatter plots for every pair of variables and univariate distributions (histograms or KDE plots) along the diagonal, making it useful for identifying correlations, patterns, clusters, and outliers.

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue='sex')
```

```python
df['sex'].value_counts()
```

## Dist Plot

A distribution plot is used to visualize the distribution of a single numerical variable. It shows how the data values are spread and helps identify the shape of the distribution, central tendency, variability, skewness, and the presence of outliers. In Seaborn, `histplot()` and `displot()` are commonly used to create distribution plots, with optional Kernel Density Estimation (KDE) curves for smoother visualization.

```python
sns.displot(data=df, x='tip', kde=True, bins=20, hue='sex')
```

```python
sns.displot(data=df, x='tip', kde=True, color='g', bins=10)
```

## Categorical Plots

- Box plot
- Violin plot
- Count plot
- Bar plot

## Count Plot

Shows the count of a variable.

```python
sns.countplot(df, x='day', palette='bright')
```

```python
sns.countplot(df, y='smoker', palette='pastel')
```

## Bar Plot

Shows the plot between two variables.

```python
sns.barplot(df, x='sex', y='total_bill', palette='deep')
```

## Box Plot

A box plot summarizes the data using the five-number summary — minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum. Box plots are particularly useful for comparing distributions across different categories and detecting skewness or outliers.

```python
sns.boxplot(df, x='sex', y='tip', palette='viridis')
```

```python
sns.boxplot(df, x='day', y='total_bill', palette='rainbow')
```

```python
sns.boxplot(df, x='total_bill', y='sex', hue='smoker')
```

## Violin Plot

Violin plot helps to see both the distribution of data in terms of Kernel Density Estimation (KDE) and the Box plot.

```python
sns.violinplot(x='total_bill', y='day', data=df, palette='plasma')
```
